```python
jug1_capacity = 4
jug2_capacity = 3
goal = 2
visited = []
def dfs(x, y):
    if (x, y) in visited:
        return False
    visited.append((x, y))
    if (x == goal or y == goal):
        print("Goal Achieved at state:", (x, y))
        return True
    if (jug1_capacity, y) not in visited:
        print("Rule: Fill Jug1")
        if dfs(jug1_capacity, y):
            return True
    if (x, jug2_capacity) not in visited:
        print("Rule: Fill Jug2")
        if dfs(x, jug2_capacity):
            return True
    if (0, y) not in visited:
        print("Rule: Empty Jug1")
        if dfs(0, y):
            return True
    if (x, 0) not in visited:
        print("Rule: Empty Jug2")
        if dfs(x, 0):
            return True
    transfer = min(x, jug2_capacity - y)
    new_x = x - transfer
    new_y = y + transfer
    if (new_x, new_y) not in visited:
        print("Rule: Pour Jug1 to Jug2")
        if dfs(new_x, new_y):
            return True
    transfer = min(y, jug1_capacity - x)
    new_x = x + transfer
    new_y = y - transfer
    if (new_x, new_y) not in visited:
        print("Rule: Pour Jug2 to Jug1")
        if dfs(new_x, new_y):
            return True
    return False
dfs(0, 0)
```
